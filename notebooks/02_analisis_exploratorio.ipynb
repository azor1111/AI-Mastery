```python
plt.figure(figsize=(10, 5))
plt.plot(df["mes"], df["ingresos"], marker='o', label="Ingresos")
plt.plot(df["mes"], df["gastos"], marker='o', label="Gastos")
plt.title("Ingresos vs Gastos - Mis Finanzas")
plt.xlabel("Mes")
plt.ylabel("Monto")
plt.legend()
plt.grid(True)
plt.show()
```
